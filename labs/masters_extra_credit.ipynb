## 🔢 Tie Breaker — Cut Line Prediction

You must predict the **cut line (total strokes)**.

### What is the Cut Line?

The cut line determines which players continue playing after the first two rounds.

- At the Masters, the **top 50 players (including ties)** after Round 2 make the cut  
- All other players are eliminated and do not play the final two rounds  
- The cut line is the **score of the last player(s) who make the cut**

### Example

- If the cut line is **+2**, all players at **+2 or better** advance  
- Players above +2 are cut  

For more details:  
https://golfweek.usatoday.com/story/sports/golf/majors/masters/2025/04/07/masters-2025-cut-line-rules/82941347007/

## 📈 Understanding Strokes Gained (SG)

**Strokes Gained measures performance relative to the field average.**

- **0** = average  
- **Positive** = better than the field  
- **Negative** = worse than the field  

### Example

- **+2.151** → Elite performance (gained ~2.15 strokes on the field)  
- **-1.108** → Below average (lost ~1.1 strokes to the field)  

### General Guidelines

- **+2 or higher** → Elite  
- **+1 to +2** → Strong  
- **0 to +1** → Slightly above average  
- **0 to -1** → Below average  
- **<-1** → Poor  

> **Higher is always better**

## 🏌️ Course Fit — Augusta National

Augusta National tends to favor players who:

- Hit the ball **longer than average off the tee**
- Have strong **approach play (iron accuracy and precision)**
- Can handle **very difficult greens**, making:
  - **Short game (around the green)** critical when greens are missed  
  - **Putting** extremely important

## Load Historical Masters Data

We have Masters data from 2021-2025. Let's load each year's data.

````python
# Load historical Masters data (2021-2025)
masters_2021 = pd.read_csv('https://raw.githubusercontent.com/bradleyboehmke/uc-bana-4080/refs/heads/main/data/masters_data/raw_pga_536_2021.csv')
masters_2022 = pd.read_csv('https://raw.githubusercontent.com/bradleyboehmke/uc-bana-4080/refs/heads/main/data/masters_data/raw_pga_14_2022.csv')
masters_2023 = pd.read_csv('https://raw.githubusercontent.com/bradleyboehmke/uc-bana-4080/refs/heads/main/data/masters_data/raw_pga_14_2023.csv')
masters_2024 = pd.read_csv('https://raw.githubusercontent.com/bradleyboehmke/uc-bana-4080/refs/heads/main/data/masters_data/raw_pga_14_2024.csv')
masters_2025 = pd.read_csv('https://raw.githubusercontent.com/bradleyboehmke/uc-bana-4080/refs/heads/main/data/masters_data/raw_pga_14_2025.csv')

# Combine all historical data into one dataframe
masters_historical = pd.concat([
    masters_2021,
    masters_2022,
    masters_2023,
    masters_2024,
    masters_2025
], ignore_index=True)

print(f"Historical data loaded: {len(masters_historical)} rows")
print(f"Years included: {sorted(masters_historical['year'].unique())}")
print(f"\nColumns available:")
print(masters_historical.columns.tolist())
````
